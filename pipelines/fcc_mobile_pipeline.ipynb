```python
from pyspark.sql.functions import monotonically_increasing_id
```

```python
# Read the data from the volume
df = spark.read.format("csv").option("header", "true").load("/Volumes/digital_redlining/mobile_fcc/src_files/fcc_mobile_full_data.csv")
```

```python
df.columns
```

```python
fact_geo = df.select(['geography_id', 'area_data_type', 'geography_type', 'geography_desc'])
dim_3g = df.select(['geography_id', 'mobilebb_3g_area_st_pct', 'mobilebb_3g_area_iv_pct'])
dim_4g = df.select(['geography_id', 'mobilebb_4g_area_st_pct', 'mobilebb_4g_area_iv_pct'])
dim_5g = df.select(['geography_id', 'mobilebb_5g_spd1_area_st_pct', 'mobilebb_5g_spd1_area_iv_pct', 'mobilebb_5g_spd2_area_st_pct', 'mobilebb_5g_spd2_area_iv_pct'])
```

```python
# Write data to tables
fact_geo.write.mode("overwrite").saveAsTable("digital_redlining.fcc_fixed.fact_geo")
dim_3g.write.mode("overwrite").saveAsTable("digital_redlining.fcc_fixed.dim_3g")
dim_4g.write.mode("overwrite").saveAsTable("digital_redlining.fcc_fixed.dim_4g")
dim_5g.write.mode("overwrite").saveAsTable("digital_redlining.fcc_fixed.dim_5g")
```

```python
# Basic validation and logging.
print(f"{fact_geo.count()} rows written to fact_geo")
print(f"{dim_3g.count()} rows written to dim_3g")
print(f"{dim_4g.count()} rows written to dim_4g")
print(f"{dim_5g.count()} rows written to dim_5g")
```
