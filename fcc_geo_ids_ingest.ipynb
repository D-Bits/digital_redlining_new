```python
fixed = spark.read.table("digital_redlining.fixed_fcc.fact_geo")
mobile = spark.read.table("digital_redlining.mobile_fcc.fact_geo")
provider_geo = spark.read.table("digital_redlining.provider_fcc.geo_provider")
```

```python
# fixed_ids = fixed.select('geography_id').distinct().collect()
# mobile_ids = mobile.select('geography_id').distinct().collect()
# provider_ids = provider_geo.select('geography_id').distinct().collect()

fixed_ids = [row['geography_id'] for row in fixed.select('geography_id').distinct().collect()]
mobile_ids = [row['geography_id'] for row in mobile.select('geography_id').distinct().collect()]
provider_ids = [row['geography_id'] for row in provider_geo.select('geography_id').distinct().collect()]

# # Combines ids and drop duplicates
all_ids = set(fixed_ids + mobile_ids + provider_ids) 
all_ids = list(all_ids)

all_ids
```
