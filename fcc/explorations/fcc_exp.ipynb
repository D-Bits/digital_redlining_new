```python
df = spark.read.format("csv").option("header", "true").load("/Volumes/digital_redlining/bronze_fcc/src_files/bdc_us_fixed_broadband_provider_summary_J25_17feb2026.csv")

df = df.orderBy('technology_code')
from pyspark.sql.functions import col

df = df.withColumn("technology_code", col("technology_code").cast("int")) 
df = df.withColumn("location_count_res", col("location_count_res").cast("int")) 
df = df.withColumn("unit_count_res", col("unit_count_res").cast("int"))
df = df.withColumn("location_count_bus", col("location_count_bus").cast("int")) 
df = df.withColumn("unit_count_bus", col("unit_count_bus").cast("int")) 

display(df)
```
